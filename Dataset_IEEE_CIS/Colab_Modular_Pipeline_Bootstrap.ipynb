## Project Path Setup

Set `PROJECT_ROOT` to your modular project folder on Drive. If you synced this folder from local machine, point to that location.

```python
from pathlib import Path
import os

# 1) Resolve modular PROJECT root (must contain code + requirements)
project_candidates = [
    Path('/content/drive/MyDrive/Colab Notebooks/Dataset_IEEE_CIS'),
    Path('/content/drive/MyDrive/Dataset_IEEE_CIS'),
    Path('/content/drive/MyDrive/Interview Kickstart/Capstone/Dataset_IEEE_CIS'),
]

PROJECT_ROOT = next(
    (p for p in project_candidates if (p / 'requirements.txt').exists() and (p / 'src').exists()),
    None,
 )
if PROJECT_ROOT is None:
    raise FileNotFoundError(
        'Could not find PROJECT_ROOT with requirements.txt and src/. '
        'Update project_candidates to your modular code folder.'
    )

# 2) Resolve DATA root (must contain train_identity.csv)
data_candidates = [
    Path('/content/drive/MyDrive/Colab Notebooks/IEEE_CIS_Dataset'),
    Path('/content/drive/MyDrive/IEEE_CIS_Dataset'),
    PROJECT_ROOT / 'ieee-fraud-detection',
]
DATA_ROOT = next((p for p in data_candidates if (p / 'train_identity.csv').exists()), None)
if DATA_ROOT is None:
    raise FileNotFoundError(
        'Could not find DATA_ROOT with train_identity.csv. '
        'Update data_candidates to your dataset folder.'
    )

os.environ['FRAUD_DATA_ROOT'] = str(DATA_ROOT)
os.chdir(PROJECT_ROOT)

print('PROJECT_ROOT:', PROJECT_ROOT)
print('DATA_ROOT:', DATA_ROOT)
print('requirements.txt exists:', (PROJECT_ROOT / 'requirements.txt').exists())
print('pyproject.toml exists:', (PROJECT_ROOT / 'pyproject.toml').exists())
```

```python
import subprocess
import sys

req_file = PROJECT_ROOT / 'requirements.txt'
if not req_file.exists():
    raise FileNotFoundError(f'Missing requirements file: {req_file}')

subprocess.check_call([sys.executable, '-m', 'pip', 'install', '--upgrade', 'pip'])
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-r', str(req_file)])

# Editable install is only valid for a Python package project root.
if (PROJECT_ROOT / 'pyproject.toml').exists() or (PROJECT_ROOT / 'setup.py').exists():
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-e', str(PROJECT_ROOT)])
    print('Editable install completed.')
else:
    print('Skipping editable install: no pyproject.toml/setup.py in PROJECT_ROOT.')

print('Dependency installation completed.')
```
